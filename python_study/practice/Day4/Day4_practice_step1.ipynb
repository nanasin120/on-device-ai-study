## 0단계. 데이터 준비 (수정하지 마세요)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    'Time': [1, 2, 3, 4, 5, 6],
    'Sensor': ['S1', 'S2', 'S1', 'S2', 'S1', 'S2'],
    'Temp': [24.5, 31.2, np.nan, 33.0, 25.0, 30.8],
    'Humid': [60, 55, 62, 50, 61, 52]
})
print("DataFrame 준비 완료:\n", df)
```

## 1단계. NumPy 배열 생성과 벡터 연산

```python
def calc_fahrenheit_mean(temp_list):
    """리스트를 numpy 배열로 만들고 화씨(C * 9/5 + 32)로 벡터 연산 후 평균값 반환"""
    arr_c = np.______(temp_list)            # ndarray 생성
    arr_f = arr_c * (9 / 5) + 32            # 벡터 사칙연산
    return round(float(np.______(arr_f)), 2) # 평균 계산

print("[1단계] 결과 :", calc_fahrenheit_mean([20.0, 30.0]))
print("[1단계] 기대값: 77.0")
```

## 3단계. Pandas 결측치 채우기 (fillna)

```python
def fill_missing_temp(data_df):
    """Temp 열의 결측치(NaN)를 Temp 열의 평균값으로 채운 데이터프레임 반환"""
    temp_mean = data_df['Temp'].______()     # 평균 계산
    df_clean = data_df.copy()
    # fillna를 사용하여 결측치를 temp_mean으로 채우세요.
    df_clean['Temp'] = df_clean['Temp'].______(temp_mean)
    return df_clean

cleaned_df = fill_missing_temp(df)
print("[3단계] 결과 :", int(cleaned_df['Temp'].isna().sum()), round(float(cleaned_df.loc[2, 'Temp']), 1))
print("[3단계] 기대값: 0 28.9")
```

## 4단계. Pandas groupby 센서별 평균 집계

```python
def group_mean_by_sensor(clean_df):
    """센서(Sensor)별로 묶어 Temp와 Humid의 평균값을 계산한 DataFrame 반환"""
    # groupby와 mean을 연결하여 집계하세요.
    grouped = clean_df.______('Sensor')[['Temp', 'Humid']].______()
    return grouped

grouped_df = group_mean_by_sensor(cleaned_df)
print("[4단계] 결과 :", grouped_df['Temp'].round(1).to_dict())
print("[4단계] 기대값: {'S1': 26.1, 'S2': 31.7}")
```

## 5단계. Matplotlib 시계열 선 그래프와 기준선 (axhline)

```python
def plot_temp_trend(clean_df, limit=30.0):
    """Time에 따른 Temp 변화 선 그래프와 limit 빨간 기준선을 그리고 fig 반환"""
    fig, ax = plt.subplots(figsize=(6, 3))
    # Time과 Temp 선 그래프 그리기
    ax.______(clean_df['Time'], clean_df['Temp'], marker='o', label='Temp')
    # 기준선 그리기 (axhline)
    ax.axhline(limit, color='red', linestyle='--', label=f'Limit ({limit}°C)')
    ax.set_title("Temperature Trend")
    ax.legend()
    return fig

fig1 = plot_temp_trend(cleaned_df)
print("[5단계] 결과 :", fig1.axes[0].get_title(), len(fig1.axes[0].lines))
print("[5단계] 기대값: Temperature Trend 2")
```
