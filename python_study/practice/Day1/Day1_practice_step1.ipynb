## 0단계. 데이터 준비 (수정하지 마세요)

## 1단계. 문자열 공백 제거 및 분리

```python
def parse_line(line):
    """'S1,24.5,60,0x01' -> ('S1', '24.5', '60', '0x01')"""
    # 양쪽 공백을 제거하고 쉼표(,)로 분리하세요.
    parts = line.______().______(',')
    return parts[0], parts[1], parts[2], parts[3]

print("[1단계] 결과 :", parse_line("S1,24.5,60,0x01"))
print("[1단계] 기대값: ('S1', '24.5', '60', '0x01')")
```

## 3단계. 비트 마스크 연산 (플래그 확인)

```python
def check_battery_low(flag_hex):
    """0x01 비트가 켜져 있으면 배터리 부족(True), 꺼져 있으면 False 반환"""
    flag = int(flag_hex, 16)       # 16진수 문자열 -> 정수
    BATTERY_MASK = 0x01            # 1번 비트 마스크
    # 비트 AND(&) 연산으로 BATTERY_MASK 비트가 켜져 있는지 확인하세요.
    return (flag ______ BATTERY_MASK) != 0

print("[3단계] 결과 :", check_battery_low("0x01"), check_battery_low("0x02"))
print("[3단계] 기대값: True False")
```

## 4단계. 조건문 판정 (온도 상태 구분)

```python
def evaluate_temp(temp):
    """32.0 이상이면 '위험', 28.0 이상이면 '주의', 그 미만이면 '정상' 반환"""
    if temp >= 32.0:
        return "______"
    elif temp >= 28.0:
        return "______"
    else:
        return "______" 

print("[4단계] 결과 :", evaluate_temp(24.5), evaluate_temp(28.0), evaluate_temp(33.5))
print("[4단계] 기대값: 정상 주의 위험")
```

## 5단계. 반복문과 enumerate 순회

```python
def label_readings(records):
    """enumerate로 1번부터 번호를 붙여 '[1번] S1: 정상' 형식 리스트 반환"""
    results = []
    # enumerate를 사용하여 1부터 시작하는 순번(idx)과 line을 순회하세요.
    for idx, line in ______(records, start=1):
        s_id, t_str, h_str, flag_hex = parse_line(line)
        status = evaluate_temp(float(t_str))
        results.append(f"[{idx}번] {s_id}: {status}")
    return results

print("[5단계] 결과 :", len(label_readings(lines)), label_readings(lines)[0])
print("[5단계] 기대값: 4 [1번] S1: 정상")
```

## 6단계. 종합 필터링 함수 (경고 센서 수집)

```python
def collect_alert_sensors(records, limit=30.0):
    """limit 이상인 센서 ID를 중복 없이 정렬된 리스트로 반환"""
    alert_sensors = []
    for line in records:
        s_id, t_str, h_str, flag_hex = parse_line(line)
        temp = float(t_str)
        # 온도가 limit 이상이고 센서 목록에 아직 없으면 추가하세요.
        if temp >= limit and (s_id ______ alert_sensors):
            alert_sensors.______(s_id)
    return sorted(alert_sensors)

print("[6단계] 결과 :", collect_alert_sensors(lines, limit=30.0))
print("[6단계] 기대값: ['S2', 'S3']")
```
