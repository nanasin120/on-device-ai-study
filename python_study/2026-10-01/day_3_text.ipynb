# 붕어빵으로 이해하는 클래스와 객체
프로그래밍에서 클래스와 객체는 길거리 붕어빵 장사를 떠올리면 가장 쉽습니다.

## 1. 클래스 (Class) = 붕어빵 틀 (철판)
- 정체: 붕어빵을 찍어내기 위해 쇠로 만들어둔 '틀(설계도)'입니다.

- 특징:

  - 붕어빵 틀 자체는 쇠붙이라서 우리가 먹을 수 없습니다.

  - 하지만 틀 하나만 잘 만들어두면, 붕어빵을 10개든 1,000개든 똑같은 모양으로 계속 찍어낼 수 있습니다.

  - 코드에서 class SensorReading:이라고 쓰는 게 바로 이 붕어빵 틀을 제작하는 작업입니다.

## 2. 객체 / 인스턴스 (Object / Instance) = 진짜 먹을 수 있는 붕어빵
- 정체: 틀에 반죽과 속재료를 부어서 실제로 구워져 나온 '진짜 붕어빵(실물)'입니다.

- 특징:

  - 배고플 때 진짜로 먹을 수 있고 만질 수 있는 실체입니다.

  - 같은 틀에서 나왔어도 1번 붕어빵에는 팥을 넣고(t=25.3), 2번 붕어빵에는 슈크림을 넣을 수(t=36.9) 있습니다.

  - 1번 붕어빵의 꼬리를 베어 먹는다고 해서 2번 붕어빵이 줄어들지 않듯, 각각의 붕어빵은 완전히 독립적입니다.

  - 코드에서 r1 = SensorReading(...)을 실행해서 변수에 담긴 결과물이 바로 이 진짜 붕어빵입니다.

# 한 줄 요약
클래스(붕어빵 틀): "붕어빵은 이렇게 생겨야 한다"라고 적어둔 쇠틀 (먹을 수 없음, 설계도)

객체(붕어빵): 그 틀로 직접 구워내서 손에 쥔 따끈따끈한 진짜 빵 (먹을 수 있음, 실물)

# 붕어빵으로 이해하는 객체 만들기 2단계
우리가 r = SensorReading("ESP32-01", 25.3)을 실행하면, 겉으로는 한 번에 뿅 나오는 것 같지만 사실은 2단계 분업으로 만들어집니다.

## 1단계: __ new __ (틀에서 빈 빵 찍어내기)
하는 일: 붕어빵 기계로 속이 텅 빈 밀가루 빵 껍데기를 먼저 딱 하나 찍어냅니다.

상태: 아직 팥도, 슈크림도 전혀 안 들어있습니다.

그냥 "너는 붕어빵 모양이다"라는 껍데기만 덩그러니 있는 상태입니다.

## 2단계: __ init __ (속재료 채워 넣기)
하는 일: 방금 나온 빈 빵 껍데기에 팥(node_id)과 슈크림(t)을 채워 넣는 작업입니다.

self가 뭐냐면: 바로 1단계에서 방금 구워져 나온 '그 빈 빵 껍데기'를 뜻합니다.

self.t = 25.3을 적는 순간, 비어 있던 빵 속에 비로소 온도가 쏙 들어갑니다.

# 딱 한 줄로 정리!
__ new __ = 속이 텅 빈 빵 만들기 (껍데기 생성)

__ init __ = 그 빵 안에 팥 채워 넣기 (내용물 채우기)

self = 팥을 넣으려고 손에 들고 있는 바로 그 빵껍데기!

### 🐟 붕어빵 주방으로 완벽 이해하는 클래스와 `self`의 비밀

파이썬의 클래스를 처음 배울 때 가장 헷갈리는 질문:
> **"왜 함수 만들 때마다 괄호 맨 앞에 `self`를 무조건 써야 할까?"**

이 질문은 길거리 **붕어빵 장사 3단계 공식**만 기억하면 완전히 풀립니다!

---

#### 📌 딱 3줄 공식 기억하기
1. **`__new__`** = 속이 텅 빈 밀가루 빵 껍데기 찍어내기 (공간 생성)
2. **`__init__`** = 그 빵 안에 팥과 슈크림 채워 넣기 (내용물 초기화)
3. **`self`** = 팥을 넣으려고 손에 들고 있는 **바로 그 빵 껍데기!**

---

#### 1. 왜 `__init__` 맨 앞자리에 `self`가 들어갈까?

붕어빵 주방 직원이 팥을 넣으려고 주걱을 들었습니다.  
그런데 **어떤 빵에 재료를 넣어야 할지** 손에 빵을 쥐고 있지 않으면 팥을 허공에 뿌리게 되겠죠?

```python
def __init__(self, node_id, t):
    self.node_id = node_id  # 이 빵에 이름표 붙이기
    self.t = t              # 이 빵에 온도 채워 넣기
```

- 주방 직원`__init__`에게 손님이 주문한 데이터(node_id, t)를 건네주면서,

- "자, 방금`__new__`가 구워낸 바로 이 빈 빵껍데기(self)에 재료를 채워 넣어라!" 하고 빵을 손에 꼭 쥐여주는 것입니다.

- 그래서 빵 채우기 작업의 1순위 준비물은 항상 그 빵 자신(self)이어야 합니다.

#### 2. 왜 다른 모든 함수(메서드)에도 self가 들어갈까?
붕어빵을 다 만들고 진열해 뒀는데, 손님이 와서 묻습니다.

"사장님, 이 붕어빵 지금 뜨거운가요? (경고 상태인가요?)"

```python
def is_alert(self):
    return self.t > 35
```

진열대에는 1번 손님 빵(r1), 2번 손님 빵(r2) 등 수많은 붕어빵이 놓여 있습니다.

"뜨겁냐"는 질문에 답하려면 주방 직원이 허공을 만져볼 수는 없습니다.

"지금 손님이 가리킨 바로 그 붕어빵(self)을 손에 집어 들어서, 그 빵의 온도(self.t)를 만져봐야" 대답할 수 있습니다.

#### 3. 코드로 보는 self의 정체 (마법의 배달 서비스)

```python
# 1. 빵 2개를 구움
r1 = SensorReading("1번빵", 25.3)
r2 = SensorReading("2번빵", 38.0)

# 우리가 쓰는 편한 코드
r1.is_alert()

# 파이썬이 뒤에서 실제로 실행하는 진짜 모습
SensorReading.is_alert(r1)
```

우리가 r1.is_alert()라고 부르면, 파이썬은 주방 매뉴얼(SensorReading.is_alert)을 펼쳐서 "1번 붕어빵(r1) 집어 들고 35도 넘는지 검사해!" 하고 첫 번째 자리(self)에 r1을 알아서 쏙 배달해 줍니다.

#### 💡 최종 요약

- 빵에 팥을 채울 때도(`__init__`)

- 다 구워진 빵의 상태를 확인할 때도(is_alert)

- "지금 다루고 있는 바로 그 빵!"이 무엇인지 알아야 하므로, 모든 함수의 첫 번째 자리는 항상 self(그 빵껍데기)가 차지합니다.

### 🏪 10초 요약: 인스턴스 변수 vs 클래스 변수

* **인스턴스 변수 (`self.t`) = 각 붕어빵의 속재료**  
  * 빵마다 다름 (1번 빵은 25도, 2번 빵은 38도).

* **클래스 변수 (`ALERT_T = 35`) = 가게 벽에 걸린 공용 칠판**  
  * 모든 빵이 공유하는 기준값.
  * 칠판 글씨 하나만 고치면(`SensorReading.ALERT_T = 30`), 구워진 모든 빵의 기준이 한 번에 싹 바뀜!

### 👨‍👦 10초 요약: 상속과 `super()` (기본 붕어빵 vs 스페셜 붕어빵)

* **상속 (`class 새클래스(옛날클래스):`)**  
  * 옛날 붕어빵 틀(부모)의 모양과 기술을 **그대로 물려받아 새 틀(자식)을 만드는 것**.  
  * 처음부터 새로 만들 필요 없이, 차이점만 덧붙이면 끝!

* **`super().__init__(...)` = "기본 조립은 부모님 공장에 맡기기"**  
  * 새 빵을 구울 때 빵 굽기와 팥 넣기 같은 **기본 작업은 부모님(`super()`)께 먼저 시키고**,  
  * 그 위에 치즈나 초코 토핑(`self.offset`)만 내가 직접 얹는 것!

### 🚪 `with open(...)`은 왜 쓸까? (자동 문지기)

파일 다루기는 **"책상 서랍을 열고 ➔ 메모지를 넣고 ➔ 서랍을 닫는 과정"**입니다.

---

#### 1. `with`를 안 쓰면 생기는 일 (`close()` 깜빡의 비극)

`with` 없이 옛날 방식으로 쓰면 반드시 마지막에 손으로 서랍을 닫아야(`f.close()`) 합니다.

```python
# 옛날 방식 (위험!)
f = open("sensor_log.txt", "w", encoding="utf-8")
f.write("T=25.3,H=60\n")
# 만약 여기서 에러가 터지거나 깜빡하고 close()를 안 쓰면?
f.close()

```

* **글이 안 적힘 (내용 증발)**: 컴퓨터는 글을 디스크에 바로 안 쓰고 임시 바구니(버퍼)에 모아뒀다가 `close()` 할 때 한 방에 쏟아 넣습니다. 닫기를 안 하면 파일에 글이 안 남고 텅 빈 파일이 될 수 있습니다.
* **파일 잠김**: 프로그램이 서랍을 쥐고 안 놔줘서, 다른 프로그램이나 엑셀에서 "이 파일이 사용 중이라 열 수 없습니다" 에러를 뿜습니다.

---

#### 2. `with`의 동작 과정 (비유: 자동문)

`with`는 파이썬이 알아서 열고 닫아주는 **자동 문지기** 역할을 합니다.

1. **들어갈 때 (`__enter__`)**: 서랍을 찰칵 열고, 그 서랍을 변수 `f`라는 이름으로 넘겨줍니다 (`as f`).
2. **안에서 일할 때**: 들여쓰기 된 블록 안에서 마음껏 글을 씁니다 (`f.write`).
3. **나갈 때 (`__exit__`)**: 들여쓰기가 끝나는 그 순간, **컴퓨터가 알아서 `close()`를 실행해 서랍을 완벽하게 잠급니다.** 중간에 에러가 터져 튕겨 나가도 무조건 닫아줍니다!

### 📦 JSON vs 📊 CSV (10초 완벽 정리)

파일에 데이터를 저장할 때 사용하는 대표적인 두 가지 양식입니다.

* **JSON (`.json`) = "파이썬 딕셔너리 그대로 박제하기"**
  * 생김새: `{"id": "ESP32", "temp": 25.3, "sub": [1, 2]}`
  * 장점: 파이썬의 `dict`, `list`, 숫자, 불리언 형태를 **원형 그대로 완벽 복원** 가능!
  * 단점: 괄호나 따옴표가 많아 데이터가 무거워짐.

* **CSV (`.csv`) = "엑셀 표 한 줄씩 적기"**
  * 생김새: `ESP32,25.3,60` (쉼표로 구분된 한 줄)
  * 장점: 가볍고 단순해서 대용량 센서 로그나 **엑셀에서 바로 열어보기**에 최적!
  * 단점: 표 형태(행·열)만 가능해서 복잡한 중첩 구조는 못 담음.

---

#### 📌 JSON 핵심 함수 4개 구분법 (`s`가 붙고 안 붙고의 차이!)

* **`json.dump(데이터, 파일)`**: 파일 객체(`f`)에 **직접 쏟아붓기 (파일 저장)**
* **`json.load(파일)`**: 파일 객체(`f`)에서 **직접 퍼오기 (파일 불러오기)**

* **`json.dumps(데이터)`**: `s` = **String(문자열)**! 파일 말고 **파이썬 글자(`str`)로 포장**
* **`json.loads(글자)`**: 글자(`str`)를 읽어서 **파이썬 딕셔너리로 언박싱**

> **외우는 팁**: 뒤에 `s`가 붙으면 메모리 속 **글자(String)**, 안 붙으면 하드디스크 **파일(File)**!

---

#### 📌 CSV 핵심: `csv.writer`

* **`w = csv.writer(f)`**: 엑셀 전용 **도장 기계** 준비
* **`w.writerow(["id", "temp", "hum"])`**: 리스트를 던져주면 알아서 쉼표(`,`) 찍고 엔터 쳐서 **표 한 줄(Row) 완성**!
* ⚠️ `open(..., newline="")`: 윈도우에서 줄 사이에 빈 줄 생기는 버그 방지용 필수 옵션!

json.dumps는 **'전 세계 모든 컴퓨터가 알아먹을 수 있는 세계 공용 표준 양식(JSON 규격)으로 str을 뽑아내는 함수'**

```python
data = {"temp": 25.3, "is_active": True}

# 1. 그냥 파이썬 글자로 바꾸기: str()
print(str(data))
# 결과: {'temp': 25.3, 'is_active': True}
# ❌ 파이썬 전용 작은따옴표(')와 대문자 True가 들어감 -> 다른 언어/웹에서 해석 불가!

# 2. JSON 규격에 맞춘 글자로 바꾸기: json.dumps()
print(json.dumps(data))
# 결과: {"temp": 25.3, "is_active": true}
# ⭕ 전 세계 표준인 큰따옴표(")와 소문자 true로 완벽 변환 -> 전 세계 어디서든 해석 가능!
```

### 🔌 시스템 연동: 센서 통신의 현실과 대처법 (1분 요약)

하드웨어(ESP32, 아두이노)나 컴퓨터 시스템과 통신할 때 맞닥뜨리는 3가지 필수 기술입니다.

---

#### 1. `Path` — 운영체제(OS) 안 가리는 똑똑한 경로
* **문제:** 윈도우는 역슬래시(`\`), 리눅스/맥은 슬래시(`/`)를 써서 경로 코드가 자주 깨짐.
* **해결:** `Path("폴더") / "파일.txt"` 처럼 **나누기(`/`) 기호**로 이어 붙이면 끝!
* 파이썬이 알아서 현재 운영체제 규격에 맞게 찰떡같이 변환해 줌.

---

#### 2. 센서 통신의 3가지 현실 (가짜 시리얼 리플레이)
전깃줄로 날아오는 센서 데이터는 절대 완벽하지 않음! 프로그램이 멈추지 않게 3가지로 걸러냄:

1. **빈 줄 (`if not line.strip(): continue`)**
   * 센서가 아직 데이터를 안 쏜 타임아웃 상태. 고장이 아니니 에러 내지 말고 부드럽게 **건너뜀**.
2. **깨진 데이터 (`except (FrameError, ValueError): continue`)**
   * 전깃줄 노이즈로 데이터가 잘려 들어옴(`$T=2`). 불량 개수 세고(`bad += 1`) 다음 줄로 **계속 진행**.
3. **정상 데이터**
   * 비로소 파싱해서 수집 완료(`good += 1`)!

> **핵심:** "에러가 안 나는 완벽한 코드를 짜는 게 아니라, **깨진 놈과 빈 놈을 유연하게 넘기며 안 멈추는 코드를 짜는 것!**"

---

#### 3. `KeyboardInterrupt` — 무한 루프에서 Ctrl + C로 안전 탈출
* 센서 수집은 24시간 돌아야 해서 `while True:` 무한 루프로 실행함.
* 끄고 싶을 때 사람이 터미널에서 **`Ctrl + C`**를 누르면 파이썬은 **`KeyboardInterrupt` 에러**를 뿜음.
* 이 에러를 `try-except`로 낚아채서 루프만 빠져나오면(`break`), **수집 중이던 파일 안전하게 저장하고 프로그램 정상 종료** 가능!

#### 📌 필수 코드 형태

```python
import subprocess

# ping 명령어로 구글 서버에 신호 1번 쏴보기
res = subprocess.run(
    ["ping", "-c", "1", "8.8.8.8"],  # (윈도우는 "-n")
    capture_output=True,  # 1. 화면에 출력하지 말고 결과 종이를 손에 쥐어 와라!
    text=True,  # 2. 0101 외계어(bytes) 말고 글자(str)로 번역해 와라!
)

print("명령어 성공 여부 (0이면 성공):", res.returncode)
print("터미널이 대답한 내용:\n", res.stdout)

```

---

#### 💡 10초 핵심 암기: 왜 저 2개 옵션을 쓸까?

1. **`capture_output=True`**
* 안 쓰면: 터미널 결과를 모니터 화면에 그냥 흩뿌려버리고 끝남.
* 쓰면: 그 결과 글자를 `res.stdout` 바구니에 담아와서 **파이썬 변수로 요리할 수 있게 해줌!**


2. **`text=True`**
* 안 쓰면: 컴퓨터 기계어 형태인 `b'Pinging 8.8.8.8...'` (바이트)로 가져와서 다루기 불편함.
* 쓰면: 우리가 읽을 수 있는 일반 파이썬 글자(`str`)로 알아서 번역해 옴!


3. **`res.returncode == 0`**
* 컴퓨터 세계의 암묵적 룰: **"0은 성공(에러 0개), 0이 아니면 실패!"**
