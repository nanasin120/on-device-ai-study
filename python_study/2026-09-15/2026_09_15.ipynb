# [Day 1 - Problem 1] 주화 자판기 거스름돈 계산 함수 구현

### 문제 설명
자판기 관리 시스템에 탑재할 **거스름돈 자동 계산 함수 `solve_change()`** 를 작성합니다.  
고객이 투입한 금액과 물건의 가격을 받아, 거슬러 줄 **동전의 총개수가 최소**가 되도록 각 동전 단위(`500`, `100`, `50`, `10`원)별 개수를 딕셔너리로 묶어 반환해야 합니다.

---

### 요구사항 및 제약조건
1. **함수 시그니처:**
   * 함수 이름: `solve_change(inserted_money, item_price)`
   * 반환값: `(coins_dict, remain_money)` 형태의 튜플 반환
     * `coins_dict`: 지급된 동전 단위와 개수를 담은 딕셔너리 (예: `{500: 7, 100: 2, 10: 3}`)
     * `remain_money`: 계산 후 최종 남은 거스름돈 정수
2. **연산자 활용:**
   * 각 동전 단위별 개수는 정수 몫 연산자(`//`)를 사용해 구합니다.
   * 다음 단위로 넘겨줄 남은 금액은 나머지 연산자(`%`)를 사용해 갱신합니다.
3. **0개 동전 처리:**
   * 지급할 동전 개수가 0개인 단위는 딕셔너리에 포함하지 않습니다. (오직 1개 이상 지급되는 단위만 저장)
4. **반복문 활용:**
   * 동전 단위 리스트 `coins = [500, 100, 50, 10]`를 `for ... in` 구문으로 순회하며 처리합니다.

---

### 입출력 예시

| 케이스 | 투입 금액 (`inserted_money`) | 물건 가격 (`item_price`) | 반환 결과 (`coins_dict`, `remain_money`) |
| :--- | :--- | :--- | :--- |
| **Case 1** | `10000` | `6270` | `({500: 7, 100: 2, 10: 3}, 0)` |
| **Case 2** | `5000` | `3500` | `({500: 3}, 0)` |
| **Case 3** | `3000` | `3000` | `({}, 0)` |
| **Case 4** | `2000` | `340` | `({500: 3, 100: 1, 50: 1, 10: 1}, 0)` |

## 정답 확인

```python
# --- 자동 채점 시스템 (수정 금지) ---
test_cases = [
    {
        "name": "Case 1 (일반 혼합 거스름돈)",
        "inserted": 10000,
        "price": 6270,
        "expected_coins": {500: 7, 100: 2, 10: 3},
        "expected_remain": 0,
    },
    {
        "name": "Case 2 (단일 동전 집중 지급)",
        "inserted": 5000,
        "price": 3500,
        "expected_coins": {500: 3},
        "expected_remain": 0,
    },
    {
        "name": "Case 3 (거스름돈 없음)",
        "inserted": 3000,
        "price": 3000,
        "expected_coins": {},
        "expected_remain": 0,
    },
    {
        "name": "Case 4 (모든 동전 단위 사용)",
        "inserted": 2000,
        "price": 340,
        "expected_coins": {500: 3, 100: 1, 50: 1, 10: 1},
        "expected_remain": 0,
    },
]

passed_count = 0

for tc in test_cases:
    actual_coins, actual_remain = solve_change(tc["inserted"], tc["price"])
    is_coins_correct = actual_coins == tc["expected_coins"]
    is_remain_correct = actual_remain == tc["expected_remain"]

    if is_coins_correct and is_remain_correct:
        print(f"✅ 통과 | {tc['name']}")
        passed_count += 1
    else:
        print(f"❌ 실패 | {tc['name']}")
        print(f"   - 반환된 결과: 동전={actual_coins}, 남은 금액={actual_remain}")
        print(f"   - 기대한 정답: 동전={tc['expected_coins']}, 남은 금액={tc['expected_remain']}")

print("\n" + "=" * 40)
if passed_count == len(test_cases):
    print(f"🎉 축하합니다! 모든 테스트 케이스({passed_count}/{len(test_cases)})를 통과했습니다.")
else:
    print(f"⚠️ {len(test_cases)}개 중 {passed_count}개 통과했습니다. 코드를 다시 점검해보세요.")
print("=" * 40)
```
