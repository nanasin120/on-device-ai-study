# [Day 1 - 문제 2] 정상 센서 로그 필터링과 평균 계산기

여러 줄로 들어온 문자열 로그에서 잘못된 형식의 줄은 건너뛰고, 정상적인 데이터만 골라내어 온도 평균과 패킷 개수를 정리하는 함수 process_sensor_logs(raw_stream)를 작성하세요.


## 1. 참고할 강의 내용

5-3: 반복문에서 특정 조건일 때 다음 회전으로 넘어가기 (continue)

6-1: 문자열의 양 끝 특정 글자 잘라내기 ([1:-1])

6-3 & 6-6: 특정 문자로 시작/끝나는지 검사 (startswith, endswith) 및 구분자로 쪼개기 (split)

6-4: 여러 줄로 된 문자열을 한 줄씩 나누기 (splitlines)

3-1: 몫과 나머지를 한 번에 튜플로 구하기 (divmod)

## 2. 문제 설명 및 요구사항

### 줄 단위 분리 및 공백 제거:

- 입력받은 문자열 raw_stream을 한 줄씩 쪼갭니다.

- 각 줄의 앞뒤 불필요한 공백을 제거합니다.

### 손상된 데이터 건너뛰기 (continue 필수):

- 줄이 $로 시작하고 *로 끝나는 형태($T=...*)가 아니라면, continue를 사용해 다음 줄로 바로 건너뜁니다.

### 온도 값 추출 및 누적:

- 정상 데이터의 양 끝($와 *)을 잘라낸 뒤, 쉼표(,)와 등호(=)를 기준으로 분리해 온도 값을 float 숫자로 변환합니다.

- 변환한 온도를 누적합에 더하고, 정상 패킷 개수를 1 증가시킵니다.

### 결과 반환:

- 정상 데이터의 온도 평균(소수점 둘째 자리 반올림)과, 정상 패킷 개수를 2로 나눈 (몫, 나머지) 튜플을 반환합니다.

- 반환 형식 예: (28.33, (1, 1))

```python
def process_sensor_logs(raw_stream):
    """
    여러 줄의 로그 문자열을 받아 정상 프레임만 처리하는 함수
    :param raw_stream: str, 여러 줄의 센서 로그 데이터
    :return: tuple, (온도_평균, (패킷수_몫, 패킷수_나머지))
    """
    total_temp = 0.0
    valid_count = 0

    # 1. splitlines()로 줄 단위 순회
    for line in raw_stream.splitlines():
        line = line.strip()

        # 2. startswith('$')와 endswith('*')가 아니면 continue로 건너뛰기
        if not (line.startswith("$") and line.endswith("*")):
            continue

        # 3. 양 끝 기호 슬라이싱 [1:-1] 후 온도(T) 파싱 및 float 변환
        content = line[1:-1]
        t_field = content.split(",")[0]  # "T=xx.x"
        temp = float(t_field.split("=")[1])

        total_temp += temp
        valid_count += 1

    # 4. 평균 계산 (round) 및 divmod로 (몫, 나머지) 계산 후 반환
    avg_temp = round(total_temp / valid_count, 2)
    packet_divmod = divmod(valid_count, 2)

    return avg_temp, packet_divmod
```

```python
# --- 테스트 실행 코드 ---
test_stream = """$T=25.4,H=60*
#NOISE_PACKET
$T=28.6,H=65*
  $T=31.0,H=70*
BAD_FRAME*
$T=19.0,H=45"""

avg, (q, r) = process_sensor_logs(test_stream)

# 기대값: 유효 데이터는 25.4, 28.6, 31.0 총 3개 (평균: 28.33, 3 // 2 = 1, 3 % 2 = 1)
assert avg == 28.33, f"평균 실패: 실제 {avg} != 정답 28.33"
assert (q, r) == (1, 1), f"divmod 실패: 실제 {(q, r)} != 정답 (1, 1)"

print("✅ Problem 2 통과!")
```
